````python
# =====================================================
#  CV Verification Agent (final working version)
# =====================================================

import asyncio
import json
from typing import Dict, List, Optional
from difflib import SequenceMatcher
from langchain_core.messages import HumanMessage
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_mcp_adapters.client import MultiServerMCPClient

# Prompt for extracting CV information
EXTRACT_PROMPT = """
You are a professional resume parser. Extract the following information from the CV text below and return it in JSON format.

Required fields:
- name: Full name of the candidate
- current_company: Current company (if explicitly stated as "currently at", otherwise take the most recent job's company)
- current_title: Current job title
- education: List of education entries, each with school, degree, field, start_year, end_year (null if still studying)
- location: City and country of the candidate, format "City, Country"
- skills: List of skills (strings)

Note:
- If a field is missing, set it to null.
- Convert years to integers, e.g., 2023.
- Return only the JSON, no additional text.

CV text:
{cv_text}
"""

def similar(a: str, b: str) -> float:
    return SequenceMatcher(None, str(a).lower(), str(b).lower()).ratio()

async def extract_cv_info(cv_text: str, llm) -> Dict:
    prompt = EXTRACT_PROMPT.format(cv_text=cv_text)
    response = await llm.ainvoke([HumanMessage(content=prompt)])
    content = response.content
    if '```json' in content:
        content = content.split('```json')[1].split('```')[0].strip()
    elif '```' in content:
        content = content.split('```')[1].split('```')[0].strip()
    try:
        return json.loads(content)
    except:
        print("Warning: Failed to parse CV info JSON.")
        return {}

def select_best_linkedin(cv_info: Dict, search_results: List[Dict]) -> Optional[Dict]:
    if not search_results:
        return None
    best = None
    best_score = -1
    cv_name = cv_info.get('name', '').lower()
    cv_location = cv_info.get('location', '').lower()
    for r in search_results:
        score = 0
        if r.get('match_type') == 'exact':
            score += 0.5
        elif r.get('match_type') == 'fuzzy':
            score += 0.2
        name_sim = similar(cv_name, r.get('name', ''))
        score += name_sim * 0.3
        if cv_location and r.get('location'):
            if cv_location in r['location'].lower() or r['location'].lower() in cv_location:
                score += 0.2
        if score > best_score:
            best_score = score
            best = r
    return best

def select_best_facebook(cv_info: Dict, search_results: List[Dict]) -> Optional[Dict]:
    if not search_results:
        return None
    best = None
    best_score = -1
    cv_name = cv_info.get('name', '').lower()
    cv_location = cv_info.get('location', '').lower()
    for r in search_results:
        score = 0
        if r.get('match_type') == 'exact':
            score += 0.5
        elif r.get('match_type') == 'fuzzy':
            score += 0.2
        name_sim = similar(cv_name, r.get('display_name', ''))
        score += name_sim * 0.3
        fb_location = f"{r.get('city', '')}, {r.get('country', '')}".lower()
        if cv_location and fb_location:
            if cv_location in fb_location or fb_location in cv_location:
                score += 0.2
        if score > best_score:
            best_score = score
            best = r
    return best

def compare_with_linkedin(cv_info: Dict, linkedin_profile: Dict) -> float:
    if not linkedin_profile:
        return 0.0
    score = 0.0
    total_weight = 0
    penalty = 1.0

    cv_name = cv_info.get('name', '').lower()
    li_name = linkedin_profile.get('name', '').lower()
    if cv_name and li_name:
        score += similar(cv_name, li_name) * 0.1
        total_weight += 0.1

    cv_company = cv_info.get('current_company', '').lower()
    cv_title = cv_info.get('current_title', '').lower()
    li_current_exp = [exp for exp in linkedin_profile.get('experience', []) if exp.get('is_current')]
    if li_current_exp:
        li_company = li_current_exp[0].get('company', '').lower()
        li_title = li_current_exp[0].get('title', '').lower()
        if cv_company and li_company:
            company_sim = similar(cv_company, li_company)
            score += company_sim * 0.3
            total_weight += 0.3
            if company_sim < 0.2:
                penalty *= 0.5
        if cv_title and li_title:
            title_sim = similar(cv_title, li_title)
            score += title_sim * 0.3
            total_weight += 0.3
            if title_sim < 0.2:
                penalty *= 0.5
    else:
        if linkedin_profile.get('experience'):
            last_exp = linkedin_profile['experience'][0]
            li_company = last_exp.get('company', '').lower()
            li_title = last_exp.get('title', '').lower()
            if cv_company and li_company:
                company_sim = similar(cv_company, li_company)
                score += company_sim * 0.25
                total_weight += 0.25
                if company_sim < 0.2:
                    penalty *= 0.5
            if cv_title and li_title:
                title_sim = similar(cv_title, li_title)
                score += title_sim * 0.25
                total_weight += 0.25
                if title_sim < 0.2:
                    penalty *= 0.5

    cv_edu = cv_info.get('education', [])
    li_edu = linkedin_profile.get('education', [])
    if cv_edu and li_edu:
        cv_first = cv_edu[0]
        li_first = li_edu[0]
        edu_score = 0
        edu_weight = 0
        if cv_first.get('school') and li_first.get('school'):
            school_sim = similar(cv_first['school'], li_first['school'])
            edu_score += school_sim * 0.5
            edu_weight += 0.5
        if cv_first.get('degree') and li_first.get('degree'):
            degree_sim = similar(cv_first['degree'], li_first['degree'])
            edu_score += degree_sim * 0.3
            edu_weight += 0.3
        if cv_first.get('field') and li_first.get('field'):
            field_sim = similar(cv_first['field'], li_first['field'])
            edu_score += field_sim * 0.2
            edu_weight += 0.2
        if edu_weight > 0:
            edu_final = edu_score / edu_weight
            score += edu_final * 0.25
            total_weight += 0.25
            if edu_final < 0.2:
                penalty *= 0.5

    cv_loc = cv_info.get('location', '').lower()
    li_loc = f"{linkedin_profile.get('city', '')}, {linkedin_profile.get('country', '')}".lower()
    if cv_loc and li_loc:
        loc_sim = similar(cv_loc, li_loc)
        score += loc_sim * 0.1
        total_weight += 0.1
        if loc_sim < 0.2:
            penalty *= 0.5

    if total_weight == 0:
        return 0.0
    raw_score = score / total_weight
    final_score = raw_score * penalty
    return max(0.0, min(1.0, final_score))

def compare_with_facebook(cv_info: Dict, fb_profile: Dict) -> float:
    if not fb_profile:
        return 0.0
    score = 0.0
    total_weight = 0
    penalty = 1.0

    cv_name = cv_info.get('name', '').lower()
    fb_name = fb_profile.get('display_name', '').lower()
    if cv_name and fb_name:
        name_sim = similar(cv_name, fb_name)
        score += name_sim * 0.1
        total_weight += 0.1
        if name_sim < 0.2:
            penalty *= 0.5

    cv_loc = cv_info.get('location', '').lower()
    fb_loc = f"{fb_profile.get('city', '')}, {fb_profile.get('country', '')}".lower()
    if cv_loc and fb_loc:
        loc_sim = similar(cv_loc, fb_loc)
        score += loc_sim * 0.25
        total_weight += 0.25
        if loc_sim < 0.2:
            penalty *= 0.5

    fb_edu = fb_profile.get('education')
    cv_edu = cv_info.get('education', [])
    if fb_edu and cv_edu:
        cv_first_school = cv_edu[0].get('school', '').lower()
        if cv_first_school:
            edu_sim = similar(cv_first_school, fb_edu.lower())
            score += edu_sim * 0.25
            total_weight += 0.25
            if edu_sim < 0.2:
                penalty *= 0.5

    fb_job = fb_profile.get('current_job', '').lower()
    fb_company = fb_profile.get('current_company', '').lower()
    cv_title = cv_info.get('current_title', '').lower()
    cv_company = cv_info.get('current_company', '').lower()

    if fb_job and cv_title:
        job_sim = similar(fb_job, cv_title)
        score += job_sim * 0.2
        total_weight += 0.2
        if job_sim < 0.2:
            penalty *= 0.5
    if fb_company and cv_company:
        company_sim = similar(fb_company, cv_company)
        score += company_sim * 0.2
        total_weight += 0.2
        if company_sim < 0.2:
            penalty *= 0.5

    if total_weight == 0:
        return 0.0
    raw_score = score / total_weight
    final_score = raw_score * penalty
    return max(0.0, min(1.0, final_score))

async def process_single_cv(cv_text: str, tools: List, llm_extract) -> float:
    cv_info = await extract_cv_info(cv_text, llm_extract)
    if not cv_info:
        return 0.2

    linkedin_score = 0.0
    linkedin_profile = None
    try:
        search_result_raw = await tools[3].ainvoke({'q': cv_info.get('name', ''), 'limit': 10, 'fuzzy': True})
        search_results = []
        if search_result_raw and isinstance(search_result_raw, list) and len(search_result_raw) > 0:
            first_item = search_result_raw[0]
            if 'text' in first_item:
                try:
                    search_results = json.loads(first_item['text'])
                except json.JSONDecodeError:
                    pass
        if search_results:
            best = select_best_linkedin(cv_info, search_results)
            if best:
                person_id = best['id']
                profile_raw = await tools[4].ainvoke({'person_id': person_id})
                if profile_raw and isinstance(profile_raw, list) and len(profile_raw) > 0:
                    first_item = profile_raw[0]
                    if 'text' in first_item:
                        try:
                            profile_data = json.loads(first_item['text'])
                            if isinstance(profile_data, list) and len(profile_data) == 1:
                                profile_data = profile_data[0]
                            linkedin_profile = profile_data
                            linkedin_score = compare_with_linkedin(cv_info, profile_data)
                        except json.JSONDecodeError:
                            pass
    except Exception:
        pass

    facebook_score = 0.0
    facebook_profile = None
    try:
        search_result_raw = await tools[0].ainvoke({'q': cv_info.get('name', ''), 'limit': 10, 'fuzzy': True})
        search_results = []
        if search_result_raw and isinstance(search_result_raw, list) and len(search_result_raw) > 0:
            first_item = search_result_raw[0]
            if 'text' in first_item:
                try:
                    search_results = json.loads(first_item['text'])
                except json.JSONDecodeError:
                    pass
        if search_results:
            best = select_best_facebook(cv_info, search_results)
            if best:
                user_id = best['id']
                profile_raw = await tools[1].ainvoke({'user_id': user_id})
                if profile_raw and isinstance(profile_raw, list) and len(profile_raw) > 0:
                    first_item = profile_raw[0]
                    if 'text' in first_item:
                        try:
                            profile_data = json.loads(first_item['text'])
                            if isinstance(profile_data, list) and len(profile_data) == 1:
                                profile_data = profile_data[0]
                            facebook_profile = profile_data
                            facebook_score = compare_with_facebook(cv_info, profile_data)
                        except json.JSONDecodeError:
                            pass
    except Exception:
        pass

    if linkedin_profile:
        final_score = linkedin_score
    elif facebook_profile:
        final_score = facebook_score * 0.8
    else:
        final_score = 0.2

    return max(0.0, min(1.0, final_score))

async def main():
    # Ensure tools are available
    try:
        tools
    except NameError:
        client = MultiServerMCPClient({
            "social_graph": {
                "transport": "http",
                "url": "https://ftec5660.ngrok.app/mcp",
                "headers": {"ngrok-skip-browser-warning": "true", "Accept": "application/json"}
            }
        })
        tools = await client.get_tools()

    # Print tool indices for verification
    print("Tools indices: 0:", tools[0].name, "3:", tools[3].name)

    # Create LLM for extraction (use the model that works for you)
    llm_extract = ChatGoogleGenerativeAI(
        model="gemini-2.5-flash",          # Change this to your working model
        google_api_key=GEMINI_VERTEX_API_KEY,
        temperature=0,
    )

    scores = []
    for cv in all_cvs:
        score = await process_single_cv(cv['text'], tools, llm_extract)
        scores.append(score)
        print(f"{cv['file']} score: {score:.3f}")
    return scores

# Run the main function
scores = await main()

# Evaluation
groundtruth = [1, 1, 1, 0, 0]  # Do not modify
result = evaluate(scores, groundtruth)
print(result)
````

# Evaluation code

In the test phase, you will be given 5 CV files with fixed names:

    CV_1.pdf, CV_2.pdf, CV_3.pdf, CV_4.pdf, CV_5.pdf

Your system must process these CVs and output a list of 5 scores,
one score per CV, in the same order:

    scores = [s1, s2, s3, s4, s5]

Each score must be a float in the range [0, 1], representing the
reliability or confidence that the CV is valid (or meets the task criteria).

The ground-truth labels are binary:

    groundtruth = [0 or 1, ..., 0 or 1]

Each CV is evaluated independently using a threshold of 0.5:

- If score > 0.5 and groundtruth == 1 → Full credit
- If score ≤ 0.5 and groundtruth == 0 → Full credit
- Otherwise → No credit

In other words, 0.5 is the decision threshold.

- Each CV contributes equally.
- Final score = (number of correct decisions) / 5


````python
# =====================================================
#  Evaluation code
# =====================================================

def evaluate(scores, groundtruth, threshold=0.5):
    """
    scores: list of floats in [0, 1], length = 5
    groundtruth: list of ints (0 or 1), length = 5
    """
    assert len(scores) == 5
    assert len(groundtruth) == 5

    correct = 0
    decisions = []

    for s, gt in zip(scores, groundtruth):
        pred = 1 if s > threshold else 0
        decisions.append(pred)
        if pred == gt:
            correct += 1

    final_score = correct / len(scores)

    return {
        "decisions": decisions,
        "correct": correct,
        "total": len(scores),
        "final_score": final_score
    }
````
